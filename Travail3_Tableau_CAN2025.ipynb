**Description** : Extraction du classement des groupes de la CAN 2025 (Maroc 2025) à partir du site Afrik-Foot avec Python (requests et pandas). Les six tableaux de groupes (A à F) sont rassemblés dans un seul tableau, puis exportés aux formats CSV et JSON.

**Source** : https://www.afrik-foot.com/can-classement-groupe

1. Téléchargement de la page et récupération des tableaux

```python
import requests
import pandas as pd
from io import StringIO

H = {"User-Agent": "Mozilla/5.0"}
url = "https://www.afrik-foot.com/can-classement-groupe"
r = requests.get(url, headers=H, timeout=30)
print("Statut :", r.status_code)

tables = pd.read_html(StringIO(r.text))
print(len(tables), "tableaux trouvés")
```

2. Assemblage des 6 groupes dans un seul tableau

```python
groupes = [t for t in tables if "Pts" in t.columns]
print(len(groupes), "classements de groupes")

for lettre, t in zip("ABCDEF", groupes):
    t.insert(0, "Groupe", lettre)

df_can = pd.concat(groupes, ignore_index=True)
print(len(df_can), "lignes")
df_can
```

3. Export en CSV et en JSON

```python
df_can.to_csv("can2025_classement.csv", index=False, encoding="utf-8-sig")
df_can.to_json("can2025_classement.json", orient="records", force_ascii=False, indent=2)

print(pd.read_csv("can2025_classement.csv").head(3))
print(open("can2025_classement.json", encoding="utf-8").read()[:400])
```
