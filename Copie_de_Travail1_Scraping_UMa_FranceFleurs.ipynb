```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

H = {"User-Agent": "Mozilla/5.0"}
print("Outils importés")
```

## 1. Données de l'Université de Maroua
Extraction du titre et du lien des articles de la page « Dernières informations » (5 premières pages) sur http://www.univ-maroua.cm.

```python
url = "http://www.univ-maroua.cm"
r = requests.get(url, headers=H, timeout=30)
print(r.status_code)

soup = BeautifulSoup(r.text, "html.parser")
print(soup.title.text)
print(len(soup.select("h3 a")), "titres trouvés")
```

```python
rows = []
for a in soup.select("h3 a"):
    rows.append({
        "titre": a.get_text(strip=True),
        "lien": requests.compat.urljoin(url, a["href"]),
    })

df_uma = pd.DataFrame(rows)
df_uma
```

```python
import time

rows = []
for page in range(5):
    r = requests.get(f"http://www.univ-maroua.cm/?page={page}", headers=H, timeout=30)
    soup = BeautifulSoup(r.text, "html.parser")
    for a in soup.select("h3 a"):
        rows.append({
            "page": page + 1,
            "titre": a.get_text(strip=True),
            "lien": requests.compat.urljoin("http://www.univ-maroua.cm", a["href"]),
        })
    time.sleep(1)

df_uma = pd.DataFrame(rows).drop_duplicates(subset="lien").reset_index(drop=True)
pd.set_option("display.max_colwidth", None)
print(len(df_uma), "articles récupérés")
df_uma
```

## 2. Données de France Fleurs
Extraction du nom, du prix et du lien des bouquets de la catégorie « Bouquet de fleurs de saison » sur https://www.francefleurs.com.

```python
url_ff = "https://www.francefleurs.com/127-bouquet-de-fleurs-de-saison"
r = requests.get(url_ff, headers=H, timeout=30)
print(r.status_code)

soup_ff = BeautifulSoup(r.text, "html.parser")
print(soup_ff.title.text)

produits = soup_ff.select("article.product-miniature")
print(len(produits), "produits trouvés")

if produits:
    print(produits[0].prettify()[:1500])
```

```python
import re

rows = []
for p in produits:
    # lien
    a = p.select_one("a.product-thumbnail")
    lien = a["href"] if a else None

    # nom : classe habituelle, sinon texte alt de l'image
    nom_tag = p.select_one(".product-title, h2, h3")
    if nom_tag:
        nom = nom_tag.get_text(strip=True)
    else:
        img = p.select_one("img")
        nom = img["alt"] if img and img.get("alt") else None

    # prix : classe habituelle, sinon recherche d'un montant en €
    prix_tag = p.select_one(".price")
    if prix_tag:
        prix = prix_tag.get_text(strip=True)
    else:
        m = re.search(r"\d+[.,]\d{2}\s*€", p.get_text(" ", strip=True))
        prix = m.group(0) if m else None

    rows.append({"nom": nom, "prix": prix, "lien": lien})

df_ff = pd.DataFrame(rows)
pd.set_option("display.max_colwidth", None)
print(len(df_ff), "produits")
df_ff
```

```python
df_ff = df_ff.drop_duplicates(subset="lien").reset_index(drop=True)

df_ff["prix_eur"] = (
    df_ff["prix"]
    .str.replace("€", "", regex=False)
    .str.replace(r"\s", "", regex=True)
    .str.replace(",", ".", regex=False)
    .astype(float)
)

print(len(df_ff), "produits")
print("Prix moyen :", round(df_ff["prix_eur"].mean(), 2), "€")
df_ff.sort_values("prix_eur").head(5)
```
