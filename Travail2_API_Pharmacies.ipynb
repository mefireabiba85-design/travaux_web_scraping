Récupération des pharmacies de garde de Maroua via l’API

```python
import requests
import pandas as pd
from google.colab import userdata

key = userdata.get("PHARMACIES_KEY")
BASE = "https://pharmacies.mit.cm/api"
headers = {"X-API-Key": key, "Accept": "application/json"}
print("Clé chargée :", bool(key))
```

```python
reponse = requests.get(f"{BASE}/pharmacies", headers=headers, timeout=30)
print("Statut :", reponse.status_code)
print("Tokens restants :", reponse.headers.get("X-RateLimit-Remaining"))
```

```python
donnees = reponse.json()
df_pharmacies = pd.json_normalize(donnees["data"])
print(len(df_pharmacies), "pharmacies")
print(list(df_pharmacies.columns))
df_pharmacies.head()
```

```python
souhaitees = ["name", "district", "address", "phone", "is_24_hours", "average_rating"]
colonnes = [c for c in souhaitees if c in df_pharmacies.columns]
df_pharmacies[colonnes]
```

```python
rep_garde = requests.get(f"{BASE}/pharmacies/on-duty", headers=headers, timeout=30)
print("Statut :", rep_garde.status_code)
print("Tokens restants :", rep_garde.headers.get("X-RateLimit-Remaining"))

df_garde = pd.json_normalize(rep_garde.json()["data"])
print(len(df_garde), "pharmacie(s) de garde")
df_garde[[c for c in ["name", "district", "address", "phone"] if c in df_garde.columns]]
```
